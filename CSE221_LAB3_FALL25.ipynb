```python
#include <iostream>
#include <vector>

using ll = long long;

ll merge(std::vector<int>& arr, std::vector<int>& temp, int left, int mid, int right) {
    ll inversion_count = 0;

    int i = left;
    int j = mid + 1;
    int k = left;

    while (i <= mid && j <= right) {
        if (arr[i] <= arr[j]) {
            temp[k++] = arr[i++];
        } else {
            temp[k++] = arr[j++];
            inversion_count += (ll)(mid - i + 1);
        }
    }

    while (i <= mid) {
        temp[k++] = arr[i++];
    }

    while (j <= right) {
        temp[k++] = arr[j++];
    }

    for (int p = left; p <= right; p++) {
        arr[p] = temp[p];
    }

    return inversion_count;
}

ll mergeSort(std::vector<int>& arr, std::vector<int>& temp, int left, int right) {
    ll inversion_count = 0;

    if (left < right) {
        int mid = left + (right - left) / 2;

        inversion_count += mergeSort(arr, temp, left, mid);
        inversion_count += mergeSort(arr, temp, mid + 1, right);
        inversion_count += merge(arr, temp, left, mid, right);
    }

    return inversion_count;
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int n;
    std::cin >> n;

    std::vector<int> arr(n);
    for (int i = 0; i < n; i++) {
        std::cin >> arr[i];
    }

    std::vector<int> temp(n);

    ll total_inversions = mergeSort(arr, temp, 0, n - 1);

    std::cout << total_inversions << "\n";

    for (int i = 0; i < n; i++) {
        std::cout << arr[i] << (i == n - 1 ? "" : " ");
    }
    std::cout << "\n";

    return 0;
}
```

```python
#include <iostream>
#include <vector>

using ll = long long;

const int MAX_VAL = 1000000;

std::vector<ll> bit;
int bit_size;

void update(int index, int delta) {
    if (index <= 0) return;

    for (; index < bit_size; index += index & -index) {
        bit[index] += delta;
    }
}

ll query(int index) {
    ll sum = 0;
    for (; index > 0; index -= index & -index) {
        sum += bit[index];
    }
    return sum;
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int n;
    std::cin >> n;

    std::vector<int> a(n);
    for (int i = 0; i < n; i++) {
        std::cin >> a[i];
    }

    bit_size = MAX_VAL + 2;
    bit.assign(bit_size, 0);

    ll total_count = 0;
    ll positive_elements_added = 0;

    for (int j = 0; j < n; j++) {
        ll val_j = a[j];

        ll query_val = val_j * val_j;

        ll count_gt = 0;

        if (query_val <= MAX_VAL) {
            ll count_lte = query((int)query_val);
            count_gt = positive_elements_added - count_lte;
        }

        total_count += count_gt;

        if (a[j] > 0) {
            update(a[j], 1);
            positive_elements_added++;
        }
    }

    std::cout << total_count << "\n";

    return 0;
}
```

Python

```python
MAX_VAL = 1000000

bit = []
bit_size = 0

def update(index, delta):
    if index <= 0:
        return
    while index < bit_size:
        bit[index] += delta
        index += index & -index

def query(index):
    sum_ = 0
    while index > 0:
        sum_ += bit[index]
        index -= index & -index
    return sum_

def main():
    import sys
    input = sys.stdin.readline

    n = int(input())
    a = list(map(int, input().split()))

    global bit_size, bit
    bit_size = MAX_VAL + 2
    bit = [0] * bit_size

    total_count = 0
    positive_elements_added = 0

    for val_j in a:
        query_val = val_j * val_j

        count_gt = 0

        if query_val <= MAX_VAL:
            count_lte = query(int(query_val))
            count_gt = positive_elements_added - count_lte

        total_count += count_gt

        if val_j > 0:
            update(val_j, 1)
            positive_elements_added += 1

    print(total_count)
main()
```

```python
#include <iostream>

using ll = long long;

const ll MOD = 107;

ll power(ll base, ll exponent) {
    ll result = 1;
    base %= MOD;

    while (exponent > 0) {
        if (exponent % 2 == 1) {
            result = (result * base) % MOD;
        }
        base = (base * base) % MOD;
        exponent /= 2;
    }
    return result;
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    ll a, b;
    std::cin >> a >> b;
    std::cout << power(a, b) << "\n";

    return 0;
}
```

```python
#include <iostream>

using ll = long long;

const ll MOD = 1000000007;

struct Matrix {
    ll mat[2][2];
};

Matrix multiply(Matrix B, Matrix C) {
    Matrix D;

    D.mat[0][0] = 0;
    D.mat[0][1] = 0;
    D.mat[1][0] = 0;
    D.mat[1][1] = 0;

    D.mat[0][0] = (B.mat[0][0] * C.mat[0][0] + B.mat[0][1] * C.mat[1][0]) % MOD;
    D.mat[0][1] = (B.mat[0][0] * C.mat[0][1] + B.mat[0][1] * C.mat[1][1]) % MOD;
    D.mat[1][0] = (B.mat[1][0] * C.mat[0][0] + B.mat[1][1] * C.mat[1][0]) % MOD;
    D.mat[1][1] = (B.mat[1][0] * C.mat[0][1] + B.mat[1][1] * C.mat[1][1]) % MOD;

    return D;
}

Matrix matrix_pow(Matrix A, ll X) {
    Matrix result;
    result.mat[0][0] = 1;
    result.mat[0][1] = 0;
    result.mat[1][0] = 0;
    result.mat[1][1] = 1;

    Matrix base = A;

    while (X > 0) {
        if (X % 2 == 1) {
            result = multiply(result, base);
        }

        base = multiply(base, base);
        X /= 2;
    }

    return result;
}

void solve() {
    Matrix A;
    std::cin >> A.mat[0][0] >> A.mat[0][1] >> A.mat[1][0] >> A.mat[1][1];

    ll X;
    std::cin >> X;

    Matrix result_matrix = matrix_pow(A, X);

    std::cout << result_matrix.mat[0][0] << " " << result_matrix.mat[0][1] << "\n";
    std::cout << result_matrix.mat[1][0] << " " << result_matrix.mat[1][1] << "\n";
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int T;
    std::cin >> T;

    while (T--) {
        solve();
    }

    return 0;
}
```

```python
#include <iostream>

using ll = long long;

struct Matrix {
    ll mat[2][2];
};

Matrix multiply(Matrix B, Matrix C, ll m) {
    Matrix D;

    D.mat[0][0] = (B.mat[0][0] * C.mat[0][0] + B.mat[0][1] * C.mat[1][0]) % m;
    D.mat[0][1] = (B.mat[0][0] * C.mat[0][1] + B.mat[0][1] * C.mat[1][1]) % m;
    D.mat[1][0] = (B.mat[1][0] * C.mat[0][0] + B.mat[1][1] * C.mat[1][0]) % m;
    D.mat[1][1] = (B.mat[1][0] * C.mat[0][1] + B.mat[1][1] * C.mat[1][1]) % m;

    return D;
}

Matrix matrix_pow(Matrix A, ll X, ll m) {
    Matrix result;
    result.mat[0][0] = 1;
    result.mat[0][1] = 0;
    result.mat[1][0] = 0;
    result.mat[1][1] = 1;

    Matrix base = A;

    while (X > 0) {
        if (X % 2 == 1) {
            result = multiply(result, base, m);
        }

        base = multiply(base, base, m);
        X /= 2;
    }

    return result;
}

void solve() {
    ll a, n, m;
    std::cin >> a >> n >> m;

    a %= m;

    Matrix T;
    T.mat[0][0] = 1;
    T.mat[0][1] = 1;
    T.mat[1][0] = 0;
    T.mat[1][1] = a;

    Matrix T_n = matrix_pow(T, n, m);

    ll ans = (T_n.mat[0][1] * a) % m;

    std::cout << ans << "\n";
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int T;
    std::cin >> T;

    while (T--) {
        solve();
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

std::vector<int> preorder;
std::vector<int> in_pos;

void printPostOrder(int pre_start, int pre_end, int in_start, int in_end) {
    if (pre_start > pre_end || in_start > in_end) {
        return;
    }

    int root_val = preorder[pre_start];

    int in_root_index = in_pos[root_val];

    int left_subtree_size = in_root_index - in_start;

    printPostOrder(pre_start + 1, pre_start + left_subtree_size, in_start, in_root_index - 1);

    printPostOrder(pre_start + left_subtree_size + 1, pre_end, in_root_index + 1, in_end);

    std::cout << root_val << " ";
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int n;
    std::cin >> n;

    preorder.resize(n);
    in_pos.resize(n + 1);

    for (int i = 0; i < n; i++) {
        int val;
        std::cin >> val;
        in_pos[val] = i;
    }

    for (int i = 0; i < n; i++) {
        std::cin >> preorder[i];
    }

    printPostOrder(0, n - 1, 0, n - 1);

    std::cout << "\n";

    return 0;
}
```
