Task 1

```python
#include <iostream>
#include <vector>
#include <queue>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n, m;
    if (!(cin >> n >> m)) return 0;

    vector<vector<int>> adj(n + 1);
    vector<int> in_degree(n + 1, 0);

    for (int i = 0; i < m; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
        in_degree[v]++;
    }

    queue<int> q;
    for (int i = 1; i <= n; ++i) {
        if (in_degree[i] == 0) {
            q.push(i);
        }
    }

    vector<int> result;
    while (!q.empty()) {
        int u = q.front();
        q.pop();
        result.push_back(u);

        for (int v : adj[u]) {
            in_degree[v]--;
            if (in_degree[v] == 0) {
                q.push(v);
            }
        }
    }

    if (result.size() != n) {
        cout << -1 << endl;
    } else {
        for (int i = 0; i < n; ++i) {
            cout << result[i] << (i == n - 1 ? "" : " ");
        }
        cout << endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <algorithm>
#include <queue>

using namespace std;

int process_component(int start_node, int n, const vector<vector<int>>& adj, vector<int>& color) {
    int count0 = 0;
    int count1 = 0;

    queue<int> q;
    q.push(start_node);
    color[start_node] = 0;
    count0++;

    while (!q.empty()) {
        int u = q.front();
        q.pop();

        for (int v : adj[u]) {
            if (color[v] == -1) {
                color[v] = 1 - color[u];

                if (color[v] == 0) count0++;
                else count1++;

                q.push(v);
            }
        }
    }

    return max(count0, count1);
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n, m;
    if (!(cin >> n >> m)) return 0;

    vector<vector<int>> adj(n + 1);
    for (int i = 0; i < m; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
        adj[v].push_back(u);
    }

    vector<int> color(n + 1, -1);
    long long total_max = 0;

    for (int i = 1; i <= n; ++i) {
        if (color[i] == -1) {
            total_max += process_component(i, n, adj, color);
        }
    }

    cout << total_max << endl;

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <queue>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n;
    if (!(cin >> n)) return 0;

    int x1, y1, x2, y2;
    cin >> x1 >> y1 >> x2 >> y2;

    if (x1 == x2 && y1 == y2) {
        cout << 0 << endl;
        return 0;
    }

    vector<vector<int>> dist(n + 1, vector<int>(n + 1, -1));
    queue<pair<int, int>> q;

    q.push({x1, y1});
    dist[x1][y1] = 0;

    int dx[] = {-2, -2, -1, -1, 1, 1, 2, 2};
    int dy[] = {-1, 1, -2, 2, -2, 2, -1, 1};

    while (!q.empty()) {
        pair<int, int> curr = q.front();
        q.pop();
        int cx = curr.first;
        int cy = curr.second;

        if (cx == x2 && cy == y2) {
            cout << dist[cx][cy] << endl;
            return 0;
        }

        for (int i = 0; i < 8; ++i) {
            int nx = cx + dx[i];
            int ny = cy + dy[i];

            if (nx >= 1 && nx <= n && ny >= 1 && ny <= n && dist[nx][ny] == -1) {
                dist[nx][ny] = dist[cx][cy] + 1;
                q.push({nx, ny});
            }
        }
    }

    if (dist[x2][y2] != -1) {
        cout << dist[x2][y2] << endl;
    } else {
        cout << -1 << endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <queue>

using namespace std;

pair<int, int> bfs(int start, int n, const vector<vector<int>>& adj) {
    vector<int> dist(n + 1, -1);
    queue<int> q;

    q.push(start);
    dist[start] = 0;

    int farthest_node = start;
    int max_dist = 0;

    while(!q.empty()) {
        int u = q.front();
        q.pop();

        if (dist[u] > max_dist) {
            max_dist = dist[u];
            farthest_node = u;
        }

        for (int v : adj[u]) {
            if (dist[v] == -1) {
                dist[v] = dist[u] + 1;
                q.push(v);
            }
        }
    }
    return {farthest_node, max_dist};
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n;
    if (!(cin >> n)) return 0;

    vector<vector<int>> adj(n + 1);
    for(int i = 0; i < n - 1; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
        adj[v].push_back(u);
    }

    pair<int, int> p1 = bfs(1, n, adj);
    int u = p1.first;

    pair<int, int> p2 = bfs(u, n, adj);
    int v = p2.first;
    int diameter = p2.second;

    cout << diameter << endl;
    cout << u << " " << v << endl;

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <queue>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n, m, s, q;
    if (!(cin >> n >> m >> s >> q)) return 0;

    vector<vector<int>> adj(n + 1);
    for (int i = 0; i < m; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
        adj[v].push_back(u);
    }

    queue<int> bfs_q;
    vector<int> dist(n + 1, -1);

    for (int i = 0; i < s; ++i) {
        int source_node;
        cin >> source_node;
        dist[source_node] = 0;
        bfs_q.push(source_node);
    }

    while (!bfs_q.empty()) {
        int u = bfs_q.front();
        bfs_q.pop();

        for (int v : adj[u]) {
            if (dist[v] == -1) {
                dist[v] = dist[u] + 1;
                bfs_q.push(v);
            }
        }
    }

    for (int i = 0; i < q; ++i) {
        int dest_node;
        cin >> dest_node;
        cout << dist[dest_node] << (i == q - 1 ? "" : " ");
    }
    cout << endl;

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <queue>
#include <algorithm>

using namespace std;

void solve() {
    int s, t;
    cin >> s >> t;

    if (s == t) {
        cout << 0 << endl;
        return;
    }
    if (s > t) {
        cout << -1 << endl;
        return;
    }

    vector<int> dist(t + 1, -1);
    queue<int> q;

    q.push(s);
    dist[s] = 0;

    while (!q.empty()) {
        int u = q.front();
        q.pop();

        if (u == t) {
            cout << dist[u] << endl;
            return;
        }

        int temp = u;
        vector<int> factors;
        for (int i = 2; i * i <= temp; ++i) {
            if (temp % i == 0) {
                factors.push_back(i);
                while (temp % i == 0) temp /= i;
            }
        }
        if (temp > 1) factors.push_back(temp);

        for (int x : factors) {
            if (x == u) continue;

            int v = u + x;
            if (v <= t && dist[v] == -1) {
                dist[v] = dist[u] + 1;
                q.push(v);
            }
        }
    }

    cout << -1 << endl;
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);
    int cases;
    if (cin >> cases) {
        while (cases--) {
            solve();
        }
    }
    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <string>
#include <queue>
#include <algorithm>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n;
    if (!(cin >> n)) return 0;

    vector<string> words(n);
    vector<bool> present(26, false);
    int unique_chars = 0;

    for (int i = 0; i < n; ++i) {
        cin >> words[i];
        for (char c : words[i]) {
            if (!present[c - 'a']) {
                present[c - 'a'] = true;
                unique_chars++;
            }
        }
    }

    vector<vector<int>> adj(26);
    vector<int> in_degree(26, 0);

    for (int i = 0; i < n - 1; ++i) {
        string w1 = words[i];
        string w2 = words[i+1];

        if (w1.size() > w2.size() && w1.substr(0, w2.size()) == w2) {
            cout << -1 << endl;
            return 0;
        }

        for (size_t j = 0; j < min(w1.size(), w2.size()); ++j) {
            if (w1[j] != w2[j]) {
                int u = w1[j] - 'a';
                int v = w2[j] - 'a';
                adj[u].push_back(v);
                in_degree[v]++;
                break;
            }
        }
    }

    priority_queue<int, vector<int>, greater<int>> pq;

    for (int i = 0; i < 26; ++i) {
        if (present[i] && in_degree[i] == 0) {
            pq.push(i);
        }
    }

    string result = "";
    while (!pq.empty()) {
        int u = pq.top();
        pq.pop();

        result += (char)(u + 'a');

        for (int v : adj[u]) {
            in_degree[v]--;
            if (in_degree[v] == 0) {
                pq.push(v);
            }
        }
    }

    if (result.length() != unique_chars) {
        cout << -1 << endl;
    } else {
        cout << result << endl;
    }

    return 0;
}
```
