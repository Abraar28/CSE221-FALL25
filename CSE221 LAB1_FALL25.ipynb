```python
T = int(input())
for _ in range(T):
    expr = input().strip()
    parts = expr.split()[1:]
    num1, op, num2 = parts[0], parts[1], parts[2]
    a = float(num1)
    b = float(num2)
    result = 0.0
    if op == '+':
        result = a + b
    elif op == '-':
        result = a - b
    elif op == '*':
        result = a * b
    elif op == '/':
        result = a / b
    print(f"{result:.6f}")
```

```python
def solve():
    n = int(input())
    prev = int(input())
    non_decreasing = True

    for _ in range(1, n):
        current = int(input())
        if prev > current:
            non_decreasing = False
        prev = current

    print("YES" if non_decreasing else "NO")

t = int(input())
for _ in range(t):
    solve()
```

```python
def reverse_subsegment(arr, start_idx, end_idx):
    arr[start_idx:end_idx+1] = arr[start_idx:end_idx+1][::-1]

def main():
    import sys
    input = sys.stdin.readline

    n = int(input())
    a = list(map(int, input().split()))

    sorted_a = sorted(a)

    initial_odd, initial_even = [], []
    sorted_odd, sorted_even = [], []

    for i in range(n):
        if i % 2 == 0:
            initial_odd.append(a[i])
            sorted_odd.append(sorted_a[i])
        else:
            initial_even.append(a[i])
            sorted_even.append(sorted_a[i])

    initial_odd.sort()
    initial_even.sort()

    if initial_odd != sorted_odd or initial_even != sorted_even:
        print("NO")
        return

    print("YES")
    operations = []

    for i in range(n):
        for j in range(i % 2, n - 2, 2):
            if a[j] > a[j + 2]:#23301680
                reverse_subsegment(a, j, j + 2)
                operations.append((j + 1, j + 3))

    print(len(operations))
    for op in operations:
        print(op[0], op[1])
main()
```

TASK G

```python
def solve():
    try:
        N = int(input())

        ids = list(map(int, input().split()))
        marks = list(map(int, input().split()))

        students = []
        for i in range(N):
            students.append([ids[i], marks[i]])

        swap_count = 0

        for i in range(N - 1):

            best_index = i

            for j in range(i + 1, N):

                candidate = students[j]
                current_best = students[best_index]

                if candidate[1] > current_best[1]:
                    best_index = j
                elif candidate[1] == current_best[1]:
                    if candidate[0] < current_best[0]:
                        best_index = j

            if i != best_index:
                students[i], students[best_index] = students[best_index], students[i]
                swap_count += 1

        print(f"Minimum swaps: {swap_count}")
        for student in students:
            print(f"ID: {student[0]} Mark: {student[1]}")

    except EOFError:
        pass
    except Exception as e:
        pass

try:
    T = int(input())

    for _ in range(T):
        solve()

except EOFError:
    pass
except Exception as e:
    pass
```

```python
from typing import List

class TrainInfo:
    def __init__(self, trainName: str, departureTime: str, originalLine: str, originalIndex: int):
        self.trainName = trainName
        self.departureTime = departureTime
        self.originalLine = originalLine
        self.originalIndex = originalIndex

def compare_trains(a: TrainInfo, b: TrainInfo) -> int:
    if a.trainName != b.trainName:
        return -1 if a.trainName < b.trainName else 1
    if a.departureTime != b.departureTime:
        return -1 if a.departureTime > b.departureTime else 1
    return -1 if a.originalIndex < b.originalIndex else 1 if a.originalIndex > b.originalIndex else 0

def main():
    n = int(input())
    trains: List[TrainInfo] = []

    for i in range(n):
        line = input()
        name_end = line.find(" will")
        train_name = line[:name_end]
        time_start = line.find(" at ") + 4 #23301680
        departure_time = line[time_start:]
        trains.append(TrainInfo(train_name, departure_time, line, i))

    # Sorting using the same logic as compareTrains
    trains.sort(key=lambda x: (x.trainName, -int(x.departureTime.replace(":", "")), x.originalIndex))

    for train in trains:
        print(train.originalLine)
main()
```
