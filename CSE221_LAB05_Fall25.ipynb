Task 1

```python
#include <iostream>
#include <vector>
#include <queue>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int N, M;
    cin >> N >> M;

    vector<vector<int>> graph(N + 1);

    for (int i = 0; i < M; ++i) {
        int u, v;
        cin >> u >> v;

        graph[u].push_back(v);
        graph[v].push_back(u);
    }

    queue<int> q;

    vector<bool> visited(N + 1, false);

    vector<int> traversal_order;

    int start_node = 1;

    q.push(start_node);
    visited[start_node] = true;
    traversal_order.push_back(start_node);

    while (!q.empty()) {
        int u = q.front();
        q.pop();

        for (int v : graph[u]) {
            if (!visited[v]) {
                visited[v] = true;
                q.push(v);
                traversal_order.push_back(v);
            }
        }
    }

    for (int i = 0; i < traversal_order.size(); ++i) {
        cout << traversal_order[i];
        if (i < traversal_order.size() - 1) {
            cout << " ";
        }
    }
    cout << "\n";

    return 0;
}
```

```python
#include <iostream>
#include <vector>

using namespace std;

vector<vector<int>> graph;
vector<bool> visited;
vector<int> traversal_order;

void dfs(int u) {
    visited[u] = true;
    traversal_order.push_back(u);

    for (int v : graph[u]) {
        if (!visited[v]) {
            dfs(v);
        }
    }
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int N, M;
    cin >> N >> M;

    graph.resize(N + 1);
    visited.resize(N + 1, false);

    vector<int> u_nodes(M);
    for (int i = 0; i < M; ++i) {
        cin >> u_nodes[i];
    }

    vector<int> v_nodes(M);
    for (int i = 0; i < M; ++i) {
        cin >> v_nodes[i];
    }

    for (int i = 0; i < M; ++i) {
        int u = u_nodes[i];
        int v = v_nodes[i];

        graph[u].push_back(v);
        graph[v].push_back(u);
    }

    int start_node = 1;
    dfs(start_node);

    for (int i = 0; i < traversal_order.size(); ++i) {
        cout << traversal_order[i];
        if (i < traversal_order.size() - 1) {
            cout << " ";
        }
    }
    cout << "\n";

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <queue>
#include <algorithm>

using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int N, M, S, D;
    cin >> N >> M >> S >> D;

    vector<vector<int>> graph(N + 1);

    vector<int> u_nodes(M);
    for (int i = 0; i < M; ++i) {
        cin >> u_nodes[i];
    }

    vector<int> v_nodes(M);
    for (int i = 0; i < M; ++i) {
        cin >> v_nodes[i];
    }

    for (int i = 0; i < M; ++i) {
        int u = u_nodes[i];
        int v = v_nodes[i];
        graph[u].push_back(v);
        graph[v].push_back(u);
    }

    for (int i = 1; i <= N; ++i) {
        sort(graph[i].begin(), graph[i].end());
    }

    queue<int> q;

    vector<int> distance(N + 1, -1);

    vector<int> parent(N + 1, 0);

    q.push(S);
    distance[S] = 0;

    bool found = false;

    while (!q.empty()) {
        int u = q.front();
        q.pop();

        if (u == D) {
            found = true;
            break;
        }

        for (int v : graph[u]) {
            if (distance[v] == -1) {
                distance[v] = distance[u] + 1;
                parent[v] = u;
                q.push(v);
            }
        }
    }

    if (!found) {
        cout << "-1\n";
    } else {
        cout << distance[D] << "\n";

        vector<int> path;
        int curr = D;

        while (curr != 0) {
            path.push_back(curr);
            curr = parent[curr];
        }

        reverse(path.begin(), path.end());

        for (int i = 0; i < path.size(); ++i) {
            cout << path[i];
            if (i < path.size() - 1) {
                cout << " ";
            }
        }
        cout << "\n";
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

vector<vector<int>> adj;
vector<int> subtree_size;

void dfs(int u, int p) {
    subtree_size[u] = 1;

    for (int v : adj[u]) {
        if (v != p) {
            dfs(v, u);
            subtree_size[u] += subtree_size[v];
        }
    }
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int N, R;
    cin >> N >> R;

    adj.resize(N + 1);
    subtree_size.resize(N + 1, 0);

    for (int i = 0; i < N - 1; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
        adj[v].push_back(u);
    }

    dfs(R, 0);

    int Q;
    cin >> Q;

    for (int i = 0; i < Q; ++i) {
        int X;
        cin >> X;
        cout << subtree_size[X] << "\n";
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

using namespace std;

bool dfs(int u, vector<vector<int>>& adj, vector<int>& color) {
    color[u] = 1;

    for (int v : adj[u]) {
        if (color[v] == 1) {
            return true;
        }
        if (color[v] == 0) {
            if (dfs(v, adj, color)) {
                return true;
            }
        }
    }

    color[u] = 2;

    return false;
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int N, M;
    cin >> N >> M;

    vector<vector<int>> adj(N + 1);

    for (int i = 0; i < M; ++i) {
        int u, v;
        cin >> u >> v;
        adj[u].push_back(v);
    }

    vector<int> color(N + 1, 0);

    bool hasCycle = false;

    for (int i = 1; i <= N; ++i) {
        if (color[i] == 0) {
            if (dfs(i, adj, color)) {
                hasCycle = true;
                break;
            }
        }
    }

    if (hasCycle) {
        cout << "YES\n";
    } else {
        cout << "NO\n";
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <string>
using namespace std;
int R, H;

int dfs(int r, int c, vector<string>& grid, vector<vector<bool>>& visited) {
    if (r < 0 || r >= R) {
        return 0;
    }
    if (c < 0 || c >= H) {
        return 0;
    }
    if (grid[r][c] == '#') {
        return 0;
    }
    if (visited[r][c]) {
        return 0;
    }

    visited[r][c] = true;

    int diamondsInThisCell = (grid[r][c] == 'D') ? 1 : 0;

    return diamondsInThisCell +
           dfs(r + 1, c, grid, visited) +
           dfs(r - 1, c, grid, visited) +
           dfs(r, c + 1, grid, visited) +
           dfs(r, c - 1, grid, visited);
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    cin >> R >> H;

    vector<string> grid(R);
    for (int i = 0; i < R; ++i) {
        cin >> grid[i];
    }

    vector<vector<bool>> visited(R, vector<bool>(H, false));

    int maxDiamonds = 0;

    for (int i = 0; i < R; ++i) {
        for (int j = 0; j < H; ++j) {
            if (grid[i][j] != '#' && !visited[i][j]) {
                int diamondsInComponent = dfs(i, j, grid, visited);

                if (diamondsInComponent > maxDiamonds) {
                    maxDiamonds = diamondsInComponent;
                }
            }
        }
    }

    cout << maxDiamonds << "\n";

    return 0;
}
```
