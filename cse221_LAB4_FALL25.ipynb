```python
#include <iostream>
#include <vector>

int main() {
    int N, M;
    std::cin >> N >> M;

    std::vector<std::vector<int>> matrix(N, std::vector<int>(N, 0));

    for (int i = 0; i < M; ++i) {
        int u, v, w;
        std::cin >> u >> v >> w;
        matrix[u - 1][v - 1] = w;
    }

    for (int i = 0; i < N; ++i) {
        for (int j = 0; j < N; ++j) {
            std::cout << matrix[i][j] << " ";
        }
        std::cout << std::endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

int main() {
    int N, M;
    std::cin >> N >> M;

    std::vector<int> U(M);
    std::vector<int> V(M);
    std::vector<int> W(M);

    for (int i = 0; i < M; ++i) std::cin >> U[i];
    for (int i = 0; i < M; ++i) std::cin >> V[i];
    for (int i = 0; i < M; ++i) std::cin >> W[i];

    std::vector<std::vector<std::pair<int, int>>> adj_list(N);

    for (int i = 0; i < M; ++i) {
        adj_list[U[i] - 1].push_back({V[i], W[i]});
    }

    for (int i = 0; i < N; ++i) {
        std::cout << (i + 1) << ":";
        for (const auto& pair : adj_list[i]) {
            std::cout << " (" << pair.first << "," << pair.second << ")";
        }
        std::cout << std::endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

int main() {
    int N, M;
    std::cin >> N >> M;

    std::vector<int> degrees(N + 1, 0);
    std::vector<int> U(M);
    std::vector<int> V(M);

    for (int i = 0; i < M; ++i) std::cin >> U[i];
    for (int i = 0; i < M; ++i) std::cin >> V[i];

    for (int i = 0; i < M; ++i) {
        degrees[U[i]]++;
        degrees[V[i]]++;
    }

    int odd_count = 0;
    for (int i = 1; i <= N; ++i) {
        if (degrees[i] % 2 != 0) {
            odd_count++;
        }
    }

    if (odd_count == 0 || odd_count == 2) {
        std::cout << "YES" << std::endl;
    } else {
        std::cout << "NO" << std::endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

int main() {
    int N, M;
    std::cin >> N >> M;

    std::vector<int> indegree(N + 1, 0);
    std::vector<int> outdegree(N + 1, 0);

    std::vector<int> U(M);
    for (int i = 0; i < M; ++i) std::cin >> U[i];

    std::vector<int> V(M);
    for (int i = 0; i < M; ++i) std::cin >> V[i];

    for (int i = 0; i < M; ++i) {
        outdegree[U[i]]++;
        indegree[V[i]]++;
    }

    for (int i = 1; i <= N; ++i) {
        std::cout << (indegree[i] - outdegree[i]) << " ";
    }

    std::cout << std::endl;

    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <algorithm>

int main() {
    int N;
    std::cin >> N;

    int x, y;
    std::cin >> x >> y;

    int dx[] = {-1, -1, -1, 0, 0, 1, 1, 1};
    int dy[] = {-1, 0, 1, -1, 1, -1, 0, 1};

    std::vector<std::pair<int, int>> valid_moves;

    for (int i = 0; i < 8; ++i) {
        int new_x = x + dx[i];
        int new_y = y + dy[i];

        if (new_x >= 1 && new_x <= N && new_y >= 1 && new_y <= N) {
            valid_moves.push_back({new_x, new_y});
        }
    }

    std::sort(valid_moves.begin(), valid_moves.end());

    std::cout << valid_moves.size() << std::endl;

    for (const auto& move : valid_moves) {
        std::cout << move.first << " " << move.second << std::endl;
    }

    return 0;
}
```

```python
#include <iostream>
#include <vector>

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    int N, M, K;
    std::cin >> N >> M >> K;

    std::vector<std::vector<bool>> is_occupied(N + 1, std::vector<bool>(M + 1, false));
    std::vector<std::pair<int, int>> knights(K);

    for (int i = 0; i < K; ++i) {
        std::cin >> knights[i].first >> knights[i].second;
        is_occupied[knights[i].first][knights[i].second] = true;
    }

    int dx[] = {1, 1, 2, 2, -1, -1, -2, -2};
    int dy[] = {2, -2, 1, -1, 2, -2, 1, -1};

    for (int i = 0; i < K; ++i) {
        int x = knights[i].first;
        int y = knights[i].second;

        for (int j = 0; j < 8; ++j) {
            int new_x = x + dx[j];
            int new_y = y + dy[j];

            if (new_x >= 1 && new_x <= N && new_y >= 1 && new_y <= M) {
                if (is_occupied[new_x][new_y]) {
                    std::cout << "YES" << std::endl;
                    return 0;
                }
            }
        }
    }

    std::cout << "NO" << std::endl;
    return 0;
}
```

```python
#include <iostream>
#include <vector>
#include <numeric>

const int MAX_N = 200005;
int spf[MAX_N];

void sieve() {
    for (int i = 1; i < MAX_N; ++i) spf[i] = i;
    for (int i = 2; (long long)i * i < MAX_N; ++i) {
        if (spf[i] == i) {
            for (int j = i * i; j < MAX_N; j += i) {
                if (spf[j] == j) {
                    spf[j] = i;
                }
            }
        }
    }
}

std::vector<int> get_prime_factors(int n) {
    std::vector<int> factors;
    while (n > 1) {
        int p = spf[n];
        factors.push_back(p);
        while (n % p == 0) {
            n /= p;
        }
    }
    return factors;
}

long long count_coprime(const std::vector<int>& primes, int num) {
    long long count = 0;
    int m = primes.size();

    for (int i = 0; i < (1 << m); ++i) {
        long long product = 1;
        int set_bits = 0;
        for (int j = 0; j < m; ++j) {
            if ((i >> j) & 1) {
                if (product > (long long)num / primes[j]) {
                    product = num + 1;
                    break;
                }
                product *= primes[j];
                set_bits++;
            }
        }

        if (set_bits % 2 == 1) {
            count -= num / product;
        } else {
            count += num / product;
        }
    }
    return count;
}

int main() {
    std::ios_base::sync_with_stdio(false);
    std::cin.tie(NULL);

    sieve();

    int N;
    int Q;
    std::cin >> N >> Q;

    while (Q--) {
        int X;
        long long K;
        std::cin >> X >> K;

        if (X == 1) {
            if (K > N - 1) {
                std::cout << -1 << "\n";
            } else {
                std::cout << K + 1 << "\n";
            }
            continue;
        }

        std::vector<int> primes = get_prime_factors(X);
        long long total = count_coprime(primes, N);

        if (K > total) {
            std::cout << -1 << "\n";
            continue;
        }

        int low = 1, high = N;
        int ans = -1;

        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (count_coprime(primes, mid) >= K) {
                ans = mid;
                high = mid - 1;
            } else {
                low = mid + 1;
            }
        }
        std::cout << ans << "\n";
    }

    return 0;
}
```
