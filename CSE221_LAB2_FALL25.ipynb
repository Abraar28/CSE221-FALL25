```python
import sys

def solve():
    try:
        n, s = map(int, sys.stdin.readline().split())
    except ValueError:
        return

    if n == 0:
        print(-1)
        return

    try:
        a = list(map(int, sys.stdin.readline().split()))
    except EOFError:
        return
    except ValueError:
        print(-1)#23301680
        return

    left = 0
    right = n - 1

    while left < right:
        current_sum = a[left] + a[right]

        if current_sum == s:
            print(left + 1, right + 1)
            return
        elif current_sum < s:
            left += 1
        else:
            right -= 1

    print(-1)

solve()
```

```python

import sys

def solve():
    try:
        n, x = map(int, sys.stdin.readline().split())
    except ValueError:
        return

    try:
        a = list(map(int, sys.stdin.readline().split()))
    except (EOFError, ValueError):
        return

    a_with_indices = []
    for i in range(n):
        a_with_indices.append((a[i], i + 1))

    a_with_indices.sort()

    for i in range(n - 2):
        val_i, idx_i = a_with_indices[i]

        target = x - val_i

        left = i + 1
        right = n - 1

        while left < right:
            val_l, idx_l = a_with_indices[left]#23301680
            val_r, idx_r = a_with_indices[right]
            current_sum = val_l + val_r

            if current_sum == target:
                print(idx_i, idx_l, idx_r)
                return
            elif current_sum < target:
                left += 1
            else:
                right -= 1

    print(-1)

solve()

```

```python

import sys

def solve():
    try:
        N, K = map(int, sys.stdin.readline().split())
    except ValueError:
        return

    if N == 0:
        print(0)
        return

    try:
        a = list(map(int, sys.stdin.readline().split()))
    except (EOFError, ValueError):
        a = []

    if not a:
        print(0)
        return
        #23301680
    left = 0
    max_length = 0
    window_counts = {}

    for right in range(N):
        num_right = a[right]
        window_counts[num_right] = window_counts.get(num_right, 0) + 1

        while len(window_counts) > K:
            num_left = a[left]

            window_counts[num_left] -= 1

            if window_counts[num_left] == 0:
                del window_counts[num_left]

            left += 1

        current_length = right - left + 1
        max_length = max(max_length, current_length)

    print(max_length)

solve()

```

```python

import sys
from bisect import bisect_left, bisect_right

def solve():
    try:
        n, q = map(int, sys.stdin.readline().split())
    except ValueError:
        return

    if n > 0:
        try:
            a = list(map(int, sys.stdin.readline().split()))
        except (EOFError, ValueError):
            a = []
    else:
        a = []

    answers = []

    for _ in range(q):
        try:
            x, y = map(int, sys.stdin.readline().split())
        except (EOFError, ValueError):
            continue#23301680

        start_index = bisect_left(a, x)

        end_index_exclusive = bisect_right(a, y)

        count = end_index_exclusive - start_index

        answers.append(str(count))

    sys.stdout.write("\n".join(answers) + "\n")

solve()

```
